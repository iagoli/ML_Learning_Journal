```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
# Pairplot to visualize class separation
sns.pairplot(df, hue='species')
plt.suptitle("Iris Dataset Feature Analysis", y=1.02)
plt.show()
```
