```python
# Exercise: ex14_titanic_data_plots
#14. Com o dataset Titanic, disponibilizado no package Seaborn, produza os seguintes gráficos.
#a) Um gráfico de barras que mostre a idade média dos homens, mulheres e crianças que
#embarcaram no Titanic.
#b) Um gráfico que mostre a dispersão das idades em cada grupo de pessoas: homens,
#mulheres e crianças.
#c) Um gráfico que mostre a taxa de sobrevivência dos homens, mulheres e crianças.
#d) Um gráfico que mostre a taxa de sobrevivência dos homens, mulheres e crianças, por
#classe de cabine.
#e) Um gráfico que mostre o valor médio da tarifa paga pelos homens, mulheres e
#crianças, por classe de cabine.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
```

```python
# a) Average Age
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='who', y='age', errorbar=None)
plt.title("Average Age by Group")
plt.show()
```

```python
# b) Age Dispersion
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='who', y='age')
plt.title("Age Dispersion by Group")
plt.show()
```

```python
# c) Survival Rate
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='who', y='survived', errorbar=None)
plt.title("Survival Rate by Group")
plt.show()
```

```python
# d) Survival by Class
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='class', y='survived', hue='who', errorbar=None)
plt.title("Survival Rate by Class & Group")
plt.show()
```

```python
# e) Average Fare
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='class', y='fare', hue='who', errorbar=None)
plt.title("Average Fare by Class & Group")
plt.show()
```
